## 4. Main Event


Start by setting up a small Poisson problem and a reusable GP sampler.


```python

import python.active_lib as active_lib
import numpy as np
import matplotlib.pyplot as plt
from dolfin import Function, TrialFunction, TestFunction, assemble, inner, dx, plot
from python.poisson import PoissonControl

rng = np.random.default_rng(123)

problem = PoissonControl(nx=16, ny=16, alpha=1e-3)
Q = problem.Q

# We use the same squared-exponential GP prior as active_lib.sample_m(),
# but factor its covariance once so repeated draws are cheap.
coords = Q.tabulate_dof_coordinates().reshape((-1, Q.mesh().geometry().dim()))
lengthscale = 0.20
sigma = 1.0

sqdist = np.sum((coords[:, None, :] - coords[None, :, :])**2, axis=-1)
K = sigma**2 * np.exp(-0.5 * sqdist / lengthscale**2)
L = np.linalg.cholesky(K + 1e-8 * np.eye(K.shape[0]))

def draw_control():
    m = Function(Q, name="control")
    m.vector().set_local(L @ rng.standard_normal(Q.dim()))
    m.vector().apply("insert")
    return m

m_example = draw_control()
plt.figure(figsize=(4, 3.5))
p = plot(m_example)
plt.colorbar(p)
plt.title("One input function $m$")
plt.tight_layout()
plt.show()
```

Next, do the Monte Carlo sample inputs and evaluate J(m) and its functional gradient.

```python
B = 100 # Monte Carlo sample size.

M = []       # sampled input/control functions
J = []       # scalar objective values
G = []       # functional gradients

for _ in range(B):
    m = draw_control()
    j, g = problem.evaluate(m)
    M.append(m)
    J.append(j)
    G.append(g)

J = np.asarray(J)

print(f"Collected {B} samples")
print(f"J range: [{J.min():.4g}, {J.max():.4g}]")

fig, ax = plt.subplots(1, 2, figsize=(8, 3.4))
plt.sca(ax[0])
p0 = plot(M[0])
plt.colorbar(p0)
ax[0].set_title("sample input $m$")

plt.sca(ax[1])
p1 = plot(G[0])
plt.colorbar(p1)
ax[1].set_title(r"gradient $\nabla J(m)$")

plt.tight_layout()
plt.show()
```

Now form the gram matrix and eigenanalysis.

```python

def mass_matrix(space):
    f = TrialFunction(space)
    q = TestFunction(space)
    return assemble(inner(f, q) * dx)

MQ = mass_matrix(Q)

def gram_matrix(F1, F2=None):
    if F2 is None:
        F2 = F1

    out = np.empty((len(F1), len(F2)))
    work = F2[0].vector().copy()

    for j, fj in enumerate(F2):
        work.zero()
        MQ.mult(fj.vector(), work)
        for i, fi in enumerate(F1):
            out[i, j] = fi.vector().inner(work)
    return out

Gamma = gram_matrix(G)
gram_evals, gram_evecs = np.linalg.eigh(Gamma)

# eigh returns ascending eigenvalues.
order = np.argsort(gram_evals)[::-1]
gram_evals = np.maximum(gram_evals[order], 0.0)
gram_evecs = gram_evecs[:, order]

active_evals = gram_evals / B

E = []
tol = 1e-12
for k, lam in enumerate(gram_evals):
    if lam <= tol:
        break

    # e_k = sum_i v_ik g_i / sqrt(lambda_k)
    e = Function(Q, name=f"active_eigenfunction_{k+1}")
    e.vector().zero()
    for i, g in enumerate(G):
        e.vector().axpy(gram_evecs[i, k] / np.sqrt(lam), g.vector())
    e.vector().apply("insert")
    E.append(e)

print(f"Recovered {len(E)} nonzero active eigenfunctions")
print("Leading empirical covariance eigenvalues:")
print(active_evals[:5])
```

Inspect the spectrum and leading eigendirections.

```python

fig, ax = plt.subplots(1, 3, figsize=(11, 3.2))

positive = active_evals[active_evals > 0]
positive = positive / positive[0]
ax[0].semilogy(np.arange(1, min(10, len(positive)) + 1),
               positive[:10], "o-")
ax[0].set_xlabel("index")
ax[0].set_ylabel("eigenvalue (rel)")
ax[0].set_title("Active spectrum")

plt.sca(ax[1])
p1 = plot(E[0])
plt.colorbar(p1)
ax[1].set_title(r"leading direction $e_1$")

plt.sca(ax[2])
p2 = plot(E[1])
plt.colorbar(p2)
ax[2].set_title(r"second direction $e_2$")

plt.tight_layout()
plt.show()

fraction_first = active_evals[0] / active_evals.sum()
fraction_first_two = active_evals[:2].sum() / active_evals.sum()
print(f"fraction in first direction:     {fraction_first:.3f}")
print(f"fraction in first two directions: {fraction_first_two:.3f}")
```

Project the input functions onto the active eigenfunctions.

Coloration gives value of the cost functional of a given input.

We see that it varies mostly smoothly, indicating that this 2D representation of this infinite dimensional problem is already capturing most of the interesting variation.

```python

Z = gram_matrix(M, E[:2])

fig = plt.plot(figsize=(3.6, 3.6))

sc0 = plt.scatter(Z[:, 0], Z[:,1], c=J)
ax = plt.gca()
plt.xlabel(r"$\langle m,e_1\rangle_{L^2}$")
plt.ylabel(r"$\langle m,e_2\rangle_{L^2}$")
plt.title("2D version of Poisson control problem.")
plt.colorbar(sc0, label="$J(m)$")

plt.tight_layout()
plt.show()
```
